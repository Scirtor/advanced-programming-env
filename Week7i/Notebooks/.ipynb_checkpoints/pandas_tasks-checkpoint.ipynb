```python
import pandas as pd
```

### Task 3: Creating Dataframe

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Evan"],
    "Age": [19, 22, 21, 20, 23],
    "Score": [85, 78, 92, 88, 74]
}

df = pd.DataFrame(data)
df
```

### Task 4: Data exploration

```python
# First 3 rows
print("First 3 rows:")
print(df.head(3))

# Summary statistics
print("\nSummary statistics:")
print(df.describe())

# Average score
average_score = df["Score"].mean()
print("\nAverage Score:", average_score)
```

### Data Filtering

```python
# Students older than 20
older_than_20 = df[df["Age"] > 20]
print("Students older than 20:")
print(older_than_20)

# Students with score above 80
score_above_80 = df[df["Score"] > 80]
print("\nStudents with score above 80:")
print(score_above_80)
```
