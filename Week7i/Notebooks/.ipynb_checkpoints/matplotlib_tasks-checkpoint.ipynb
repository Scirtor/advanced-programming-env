```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Sample Data
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Evan"],
    "Score": [85, 78, 92, 88, 74]
}

df = pd.DataFrame(data)
df
```

### Task 6: Line Plot

```python
plt.figure()

plt.plot(df["Name"], df["Score"], marker='o')

plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Line Plot")

plt.show()
```

### Task 7: Bar Chart

```python
plt.figure()

plt.bar(df["Name"], df["Score"])

plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Bar Chart")

plt.show()
```
