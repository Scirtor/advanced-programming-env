```python
import numpy as np
```

### Task 1: Creating Arrays

```python
arr_1_dimension = np.arange(1, 11)
```

```python
arr_2_dimension = np.random.randint(1, 11, size=(3, 3))
```

```python
print("1D Array:")
print(arr_1_dimension)

print("\n2D Array:")
print(arr_2_dimension)
```

### Task 2: Array Operations

```python
# Sum, mean, max
sum_val = arr_1_dimension.sum()
mean_val = arr_1_dimension.mean()
max_val = arr_1_dimension.max()

print("Sum:", sum_val)
print("Mean:", mean_val)
print("Max:", max_val)

# Multiply all elements by 2
multiplied = arr_1_dimension * 2
print("\nMultiplied by 2:")
print(multiplied)

# Replace values > 5 with 0
modified = arr_1_dimension.copy()
modified[modified > 5] = 0

print("\nValues > 5 replaced with 0:")
print(modified)
```
